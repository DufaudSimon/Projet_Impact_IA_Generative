```python
import pandas as pd, folium

df = pd.read_csv("export_summary.csv")
df = df[df["country"] == "France"]
df["region"] = df["region"].str.replace("\\,", "'", regex=False)
df["address"] = df["address"].str.replace("\\,", "'", regex=False)

# Tableau
cols = ["name", "operator", "city_name", "region", "power_total_mw",
        "operation_start_year", "progress_step"]
df[cols].sort_values("power_total_mw", ascending=False)
```

```python
# Carte
pts = df.dropna(subset=["latitude", "longitude"])
pts = pts[pts.latitude.between(41, 52) & pts.longitude.between(-6, 10)]

m = folium.Map(location=[46.6, 2.5], zoom_start=6, tiles="CartoDB positron")
for _, r in pts.iterrows():
    folium.CircleMarker(
        [r.latitude, r.longitude],
        radius=3 + (r.power_total_mw if pd.notna(r.power_total_mw) else 1) ** 0.5,
        color="#1f7a6d" if r.progress_step == "operating" else "#d9822b",
        fill=True, fill_opacity=0.6,
        tooltip=f"{r['name']} ({r.operator}) - {r.power_total_mw} MW",
    ).add_to(m)
m
```
